## 1. Planteamiento del problema

Se desea aproximar numéricamente la integral

$$
\int_0^2 e^{2x}\sin(3x)\,dx
$$

utilizando la Regla Compuesta de Simpson, buscando una aproximación
con un error menor que

$$
10^{-4}.
$$

## 2. Determinación de $n$ y $h$

Para la Regla Compuesta de Simpson, la cota del error está dada por

$$
|E_S|\leq \frac{b-a}{180}h^4
\max_{a\leq x\leq b}|f^{(4)}(x)|.
$$

Para este problema:

$$
f(x)=e^{2x}\sin(3x), \qquad a=0,\qquad b=2.
$$

La cuarta derivada es

$$
f^{(4)}(x)=
-e^{2x}\left(119\sin(3x)+120\cos(3x)\right).
$$

En el intervalo $[0,2]$ se obtiene aproximadamente

$$
\max |f^{(4)}(x)|\approx4475.4098.
$$

Sustituyendo en la fórmula del error:

$$
10^{-4}\geq
\frac{2}{180}(4475.4098)h^4.
$$

De donde

$$
h\leq0.0376576.
$$

Como

$$
h=\frac{b-a}{n},
$$

se obtiene

$$
n\geq53.1102.
$$

Debido a que $n$ debe ser par para la Regla Compuesta de Simpson, se toma

$$
\boxed{n=54}.
$$

Por lo tanto,

$$
h=\frac{2-0}{54}=0.037037037.
$$

Así, los parámetros que se utilizarán en la implementación son

$$
\boxed{n=54,\qquad h=0.037037037}.
$$

## 3. Algoritmo de la Regla Compuesta de Simpson

La Regla Compuesta de Simpson aproxima una integral definida
dividiendo el intervalo $[a,b]$ en $n$ subintervalos de igual tamaño,
donde $n$ debe ser un número par.

El tamaño de cada subintervalo se calcula mediante

$$
h=\frac{b-a}{n}.
$$

Los puntos utilizados son

$$
x_i=a+ih,\qquad i=0,1,\ldots,n.
$$

La aproximación de Simpson está dada por

$$
S_n=
\frac{h}{3}
\left[
f(x_0)+f(x_n)
+4\sum_{\substack{i=1\\i\text{ impar}}}^{n-1}f(x_i)
+2\sum_{\substack{i=2\\i\text{ par}}}^{n-2}f(x_i)
\right].
$$

Por lo tanto, el algoritmo debe separar las evaluaciones de la función
en índices pares e impares y posteriormente aplicar la fórmula anterior.


```python
import math
import numpy as np
import matplotlib.pyplot as plt
```

```python
def f(x):
    return math.exp(2*x) * math.sin(3*x)
```

```python
def simpson(f, a, b, n):
    
    # Verificamos que n sea un número par
    if n % 2 != 0:
        raise ValueError("n debe ser un número par")
    
    # Calculamos el tamaño de cada subintervalo
    h = (b - a) / n
    
    # Primeros y últimos términos de la fórmula
    suma = f(a) + f(b)
    
    # Sumamos los términos con índices impares
    for i in range(1, n, 2):
        x = a + i * h
        suma += 4 * f(x)
    
    # Sumamos los términos con índices pares
    for i in range(2, n, 2):
        x = a + i * h
        suma += 2 * f(x)
    
    # Aplicamos la fórmula de Simpson
    resultado = (h / 3) * suma
    
    return resultado
```

```python
a = 0
b = 2
n = 54

resultado = simpson(f, a, b, n)

print("Valor aproximado de la integral:", resultado)
```

## 4. Cálculo del error

Para comprobar la precisión de la aproximación obtenida mediante
la Regla Compuesta de Simpson, se compara el resultado numérico con
el valor de referencia de la integral.

La integral

$$
\int_0^2 e^{2x}\sin(3x)\,dx
$$

tiene como valor de referencia

$$
I\approx -14.2139771299.
$$

El error absoluto se calcula mediante

$$
E=\left|I-S_n\right|,
$$

donde $I$ representa el valor de referencia y $S_n$ la aproximación
obtenida mediante la Regla Compuesta de Simpson.

```python
valor_referencia = -14.2139771299

error = abs(valor_referencia - resultado)

print("Valor aproximado:", resultado)
print("Valor de referencia:", valor_referencia)
print("Error absoluto:", error)
```
