## Método de Integración de Romberg

La Integración de Romberg es un método numérico que permite aproximar una integral definida. El método utiliza la regla del trapecio y posteriormente mejora las aproximaciones mediante la extrapolación de Richardson.

La tabla de Romberg se representa mediante los valores $R_{i,j}$.

Para comenzar la tabla se utiliza la regla del trapecio:

$R_{1,1} = \frac{b-a}{2}[f(a)+f(b)]$

Para obtener las siguientes aproximaciones de la primera columna se utiliza:

$R_{i,1} = \frac{1}{2}R_{i-1,1} + h_i\sum f(x)$

donde el tamaño del subintervalo es:

$h_i = \frac{b-a}{2^{i-1}}$

Después se utiliza la extrapolación de Richardson para obtener las demás columnas:

$R_{i,j} = R_{i,j-1} + \frac{R_{i,j-1}-R_{i-1,j-1}}{4^{j-1}-1}$

En este ejercicio se conocen $R_{3,1}$ y $R_{3,3}$. Utilizando las relaciones de la tabla de Romberg podremos encontrar los valores intermedios y posteriormente determinar el valor desconocido de $f(2.5)$.

## Cálculo del primer elemento de la tabla de Romberg

Para comenzar la tabla de Romberg calculamos $R_{1,1}$ utilizando la regla del trapecio.

Los datos conocidos son:

$f(2)=0.51342$

$f(3)=0.36788$

Los límites de integración son:

$a=2$

$b=3$

La fórmula para calcular $R_{1,1}$ es:

$R_{1,1}=\frac{b-a}{2}[f(a)+f(b)]$

Sustituyendo los valores:

$R_{1,1}=\frac{3-2}{2}[0.51342+0.36788]$

Por lo tanto:

$R_{1,1}=0.44065$

Este valor será utilizado para calcular los siguientes elementos de la tabla de Romberg.

## Obtención de $R_{2,1}$

Para encontrar $f(2.5)$ necesitamos obtener primero el valor de $R_{2,1}$.

A partir de la fórmula de Romberg se tiene:

$R_{2,2}=R_{2,1}+\frac{R_{2,1}-R_{1,1}}{3}$

Simplificando:

$R_{2,2}=\frac{4R_{2,1}-R_{1,1}}{3}$

Para la tercera fila:

$R_{3,2}=R_{3,1}+\frac{R_{3,1}-R_{2,1}}{3}$

Simplificando:

$R_{3,2}=\frac{4R_{3,1}-R_{2,1}}{3}$

Finalmente:

$R_{3,3}=R_{3,2}+\frac{R_{3,2}-R_{2,2}}{15}$

Con los datos conocidos:

$R_{1,1}=0.44065$

$R_{3,1}=0.43687$

$R_{3,3}=0.43662$

Al sustituir estos valores y despejar $R_{2,1}$ se obtiene:

$R_{2,1}=0.4376215$

Por lo tanto:

$R_{2,1}\approx0.4376215$

## Cálculo de $f(2.5)$

Para la segunda fila de la tabla de Romberg se utiliza:

$R_{2,1}=\frac{1}{2}R_{1,1}+h_2f(2.5)$

Como:

$h_2=\frac{b-a}{2}=\frac{3-2}{2}=0.5$

sustituimos los valores conocidos:

$0.4376215=\frac{0.44065}{2}+0.5f(2.5)$

Entonces:

$0.4376215=0.220325+0.5f(2.5)$

Despejando:

$f(2.5)=\frac{0.4376215-0.220325}{0.5}$

Por lo tanto:

$f(2.5)=0.434593$

El valor buscado es:

$f(2.5)\approx0.43459$

```python
# Datos del ejercicio

a = 2
b = 3

f_a = 0.51342
f_b = 0.36788

R31 = 0.43687
R33 = 0.43662

# Cálculo de R11 mediante la regla del trapecio
R11 = ((b - a) / 2) * (f_a + f_b)

print("R11 =", R11)
print("R31 =", R31)
print("R33 =", R33)
```

```python
# Cálculo de R21 a partir de R33

R21 = (R11 / 20) + (16 * R31 / 5) - (9 * R33 / 4)

# Cálculo de los valores intermedios
R22 = (4 * R21 - R11) / 3
R32 = (4 * R31 - R21) / 3

print("R21 =", R21)
print("R22 =", R22)
print("R32 =", R32)
```

```python
# Cálculo de f(2.5)

h = (b - a) / 2

f_25 = (R21 - R11 / 2) / h

print("h =", h)
print("f(2.5) =", f_25)
```

## Implementación del algoritmo de Romberg

Para resolver el problema mediante Python se implementará el algoritmo de Romberg.

El programa utilizará los valores conocidos del ejercicio y aplicará las fórmulas de extrapolación para construir la tabla de Romberg.

El procedimiento consiste en:

1. Calcular el primer valor de la tabla mediante la regla del trapecio.
2. Obtener los valores de las siguientes filas.
3. Aplicar la extrapolación de Richardson.
4. Utilizar el valor conocido de $R_{3,3}$ para determinar $R_{2,1}$.
5. Despejar finalmente el valor de $f(2.5)$.

```python
# Integración de Romberg
# Ejercicio 2 de la Tarea 1

# Límites de integración
a = 2
b = 3

# Valores conocidos de la función
f_a = 0.51342
f_b = 0.36788

# Valores conocidos de Romberg
R31 = 0.43687
R33 = 0.43662

# Cálculo de R11 mediante la regla del trapecio
R11 = ((b - a) / 2) * (f_a + f_b)

# Cálculo de R21 a partir de R33
R21 = (R11 / 20) + (16 * R31 / 5) - (9 * R33 / 4)

# Cálculo de los valores intermedios
R22 = (4 * R21 - R11) / 3
R32 = (4 * R31 - R21) / 3

# Tamaño del subintervalo
h = (b - a) / 2

# Cálculo de f(2.5)
f_25 = (R21 - R11 / 2) / h

# Mostrar resultados
print("TABLA DE ROMBERG")
print("----------------")
print("R11 =", R11)
print("R21 =", R21)
print("R22 =", R22)
print("R31 =", R31)
print("R32 =", R32)
print("R33 =", R33)

print("\nRESULTADO")
print("---------")
print("f(2.5) =", f_25)
```
